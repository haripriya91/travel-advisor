```python
import os
from pathlib import Path
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(Path.cwd().parent / ".env")

client = OpenAI()
model = "gpt-6-luna"
```

```python
#First message
response1 = client.responses.create(
    model=model,
    instructions="You are a helpful travel advisor. Keep replies short.",
    input="I want to visit Rome for 3 days. I enjoy history."
)

print(response1.output_text)
print("Response ID:", response1.id)
```

```python
response2 = client.responses.create(
    model=model,
    instructions="You are a helpful travel advisor. Keep replies short.",
    previous_response_id=response1.id,
    input="I also enjoy Italian food. What do you recommend?"
)

print(response2.output_text)
```

```python
#See token usage
total_input = (
    response1.usage.input_tokens
    + response2.usage.input_tokens
)

total_output = (
    response1.usage.output_tokens
    + response2.usage.output_tokens
)

print("Total input tokens:", total_input)
print("Total output tokens:", total_output)
```
